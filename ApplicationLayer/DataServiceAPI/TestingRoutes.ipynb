This Notebook is made to test the DataService module. This Idea is to make multiple service calls, concurrently to test if the service can serve multiple clients. 


```python
import DataService as DataService
import asyncio
```

```python
#Generic Function for running n number of conncurrent calls where n is length of 'Params'
async def Test(Params):
    tasks = Params
    results = await asyncio.gather(*tasks)

    return results
```

Testing 'RequestLatestKPIForProgram':

```python
Params = [
    DataService.RequestLatestKPIForProgram(ProgramID=4),
    DataService.RequestLatestKPIForProgram(ProgramID=4),
    DataService.RequestLatestKPIForProgram(ProgramID=3),
    DataService.RequestLatestKPIForProgram(ProgramID=3)
]

Results = await Test(Params=Params)

idx = 1
for frame in Results:
    print(f'Results [{idx}] =>\n')
    print(frame.head(10))
    idx+=1
```
